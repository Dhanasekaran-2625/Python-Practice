```python
#single inheritance
class Employee:
    def employee_details(self):
        print("Employee Name: Arun")
        print("Employee ID: 101")

class Developer(Employee):
    def developer_details(self):
        print("Programming Language: Python")

obj = Developer()
obj.employee_details()
obj.developer_details()
```

```python
#Multiple inheritance
class student:
    def student_details(self):
        print("student name:rudhra")
        print("student id:STL198253")
class course:
    def course_details(self):
        print("course:Data analyst")
class institute(student,course):
    def institute_detail(self):
        print("Institute name : SLA")

obj = institute()

obj.student_details()
obj.course_details()
obj.institute_detail()

```

```python
# heirarchiall inheritance
class employee:
    def employee_detials(self):
        print("Emp_name : shiva")
        print("Emp_ID :STL198253")

class department(employee):
    def department_role(self):
        print("Analyst")

class experiance(employee):
    def experiance_months(self):
        print("5 years")

obj1 = department()
obj2 = experiance()


obj1.employee_detials()
obj1.department_role()
obj2.experiance_months()
```
