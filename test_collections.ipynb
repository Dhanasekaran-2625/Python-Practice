```python
shoping_cart = ["chips", "milk", "shampoo", "maggi", "apple"]

remove_item = input("Enter the item: ")

if remove_item in shoping_cart:
    shoping_cart.remove(remove_item)
    print(remove_item, "removed")
else:
    print("Item not found")

check_item = input("enter the item:")
if check_item in shoping_cart:
   print(check_item,"is found")
else:
   print(check_item,"is not found")
total_items = len(shoping_cart)
print("Total items:",total_items)
```
